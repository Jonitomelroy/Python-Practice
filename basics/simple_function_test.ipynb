```python
def addtion(a,b):
  return a+b
def subtsction(a,b):
  return a-b
def mutiply(a,b):
  return a*b
def div(a,b):
  return a/b

i=1
while i!=0:
  mode=int(input("""Enter
  1 to add
  2 to sub
  3 to mutiply
  4 to divide
  0 to exit
  :"""))

  i=mode

  if i==0:
    break


  a,b=int(input("Enter the first number:")),int(input("Enter the second number:"))
  if mode==1:
    print(addtion(a,b))
  elif mode==2:
    print(subtsction(a,b))
  elif mode==3:
    print(mutiply(a,b))
  elif mode==4:
    print(div(a,b))
  else:
    print("Invalid input")

  print("")
```
