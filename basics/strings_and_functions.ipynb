```python
name="joNitO"
great='Hello'
great1="good morning"

# Concatenation
Full_Great=great+" "+name+" "+great1
print(Full_Great)

# Use of len() function
print("total number of characters",len(Full_Great))

# slicing of stinge
print(f"First 3 latters of name:{name[:3]}")
print(f"last 3 latters of name:{name[-3:]}")

# ues of .Upper() function
print(Full_Great.upper())

# use of .lower() function
print(Full_Great.lower())

# use of .title() function
print(Full_Great.title())

# starts with
print(Full_Great.startswith("Hello"))
# ends with
print(Full_Great.endswith("Hello"))



# find the psition / find sub sting
print(Full_Great.find("jo"))
print(Full_Great.find("p"))

#replace word
print(Full_Great.replace("good","bad"))

```
