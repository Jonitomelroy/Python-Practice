```python
def fact(a):
  f=1
  for i in range (1,a+1):
    f*=i
  return f

n=10
r=4
combinations=fact(n)/(fact(r)*fact((n-r)))
print(combinations)
```

```python
import pandas as pd
import numpy as np

# -------------------------------
# Basic Statistical Analysis in Python
# -------------------------------

# Sample dataset: heights (cm) of 10 people
data = pd.DataFrame({
  'Height': [165, 170, 168, 172, 160, 175, 169, 180, 158, 166]
})

# 1. View the dataset
print("Dataset:")
print(data)

# 2. Summary statistics
print("Summary Statistics:")
print(data.describe()) # pandas describe() provides summary statistics

# 3. Mean, Median, Standard Deviation
mean_height = data['Height'].mean()
median_height = data['Height'].median()
sd_height = data['Height'].std()

print(f"\nHeight - Mean: {mean_height} Median: {median_height} SD: {sd_height}")
```
