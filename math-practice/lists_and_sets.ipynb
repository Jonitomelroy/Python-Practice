```python
import pandas as pd
import re

def file_analize(file_name):
  with open(file_name,"r")as f:text=f.read()
  alfbts=sum(c.isalpha() for c in text)
  digits=sum(c.isdigit() for c in text)
  spaces=sum(c.isspace() for c in text)
  upper=sum(c.isupper() for c in text)
  lower=sum(c.islower() for c in text)

  word=re.findall(r"\b\w+\b",text.lower())  #seperates the words in a list in individual elements
  word_starting_vowels=sum(1 for w in word if w[0] in "aeiou")
  computer_word=word.count("computer")
  data={"Alphabets":[alfbts],
        "digits":[digits],
        "spaces":[spaces],
        "upper":[upper],
        "lower":[lower],
        "words starting with vowels":[word_starting_vowels],
        "number time of computer is comeing":[computer_word]}
  df=pd.DataFrame(data)
  return df

if __name__=="__main__":
  file_name="/content/computer.txt"
  result=file_analize(file_name)
  print(result)
```
