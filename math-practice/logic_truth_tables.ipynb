```python
import math

def poisson_pmf_simple(k, lambda_val):
  return (lambda_val ** k) * math.exp(-lambda_val) / math.factorial(k)

# Example:
k_val = 2  # Number of events
lambda_val = 3.0 # Average rate of events
probability = poisson_pmf_simple(k_val, lambda_val)
print(f"P(X={k_val}) when lambda={lambda_val} is: {probability:.4f}")
```
