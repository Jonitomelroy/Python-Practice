```python
#the probability of getting exactly 5 head
def fact(a):
  if a==0:
    return 1
  else:
    return a*fact(a-1)
def ncr(n,r):
  return fact(n)/(fact(r)*fact(n-r))

n=8
k=5
p=0.5
probability=(ncr(n,k))*(p**k)*(1-p)**(n-k)
print("the probability of getting exactly 5 head",probability)
```

```python
def fibo(n):
  if n==1:
    return 1
  elif n==0:
    return 0
  else:
    return (fibo(n-1)+fibo(n-2))
fibo(9)
```
