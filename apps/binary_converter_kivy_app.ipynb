```python
#!/usr/bin/env python3
"""
Binary Converter App - Python application with Kivy GUI
Can be packaged as APK using Buildozer for Android
"""

import kivy
from kivy.app import App
from kivy.uix.boxlayout import BoxLayout
from kivy.uix.gridlayout import GridLayout
from kivy.uix.label import Label
from kivy.uix.textinput import TextInput
from kivy.uix.button import Button
from kivy.uix.popup import Popup
from kivy.core.window import Window
import re

kivy.require('2.1.0')

class BinaryConverter:
    """Binary conversion utilities"""

    @staticmethod
    def decimal_to_binary(n):
        """Convert decimal to binary"""
        b = ""
        while n != 0:
            b = str(n % 2) + b
            n = n // 2
        return b if b else "0"

    @staticmethod
    def binary_addition(b1, b2):
        """Add two binary numbers"""
        len_ = max(len(b1), len(b2))
        b1 = b1.zfill(len_)
        b2 = b2.zfill(len_)
        result = ""
        carry = 0
        for i in range(len_ - 1, -1, -1):
            bitsum = int(b1[i]) + int(b2[i]) + carry
            result = str(bitsum % 2) + result
            carry = bitsum // 2
        if carry:
            result = str(carry) + result
        return result

    @staticmethod
    def binary_to_octal(n):
        """Convert binary to octal"""
        len_ = len(n) - 1
        result = 0
        for i in n:
            result = (2 ** len_) * int(i) + result
            len_ -= 1
        n = result
        b = ""
        while n != 0:
            b = str(n % 8) + b
            n = n // 8
        return b if b else "0"

    @staticmethod
    def binary_to_hex(n):
        """Convert binary to hexadecimal"""
        hexa = '0123456789ABCDEF'
        len_ = len(n) - 1
        result = 0
        for i in n:
            result = (2 ** len_) * int(i) + result
            len_ -= 1
        n = result
        b = ""
        while n != 0:
            b = hexa[n % 16] + b
            n = n // 16
        return b if b else "0"

class BinaryConverterLayout(BoxLayout):
    """Main layout for the binary converter app"""

    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        self.orientation = 'vertical'
        self.padding = 20
        self.spacing = 15

        # Title
        title = Label(
            text='Binary Converter Pro',
            font_size='24sp',
            bold=True,
            size_hint_y=None,
            height=50,
            color=(0.2, 0.2, 0.2, 1)
        )
        self.add_widget(title)

        # Operation selection
        self.operation_spinner = Spinner(
            text='Select Operation',
            values=('Decimal to Binary', 'Binary Addition', 'Binary to Octal', 'Binary to Hexadecimal'),
            size_hint=(1, None),
            height=44,
            background_color=(0.4, 0.6, 0.8, 1)
        )
        self.operation_spinner.bind(text=self.on_operation_change)
        self.add_widget(self.operation_spinner)

        # Input fields container
        self.input_layout = BoxLayout(orientation='vertical', spacing=10)

        # First input
        self.input1_layout = BoxLayout(orientation='horizontal', size_hint_y=None, height=44)
        self.input1_layout.add_widget(Label(text='Input 1:', size_hint_x=0.3))
        self.input1 = TextInput(multiline=False, size_hint_x=0.7)
        self.input1_layout.add_widget(self.input1)
        self.input_layout.add_widget(self.input1_layout)

        # Second input (initially hidden)
        self.input2_layout = BoxLayout(orientation='horizontal', size_hint_y=None, height=44)
        self.input2_layout.add_widget(Label(text='Input 2:', size_hint_x=0.3))
        self.input2 = TextInput(multiline=False, size_hint_x=0.7)
        self.input2_layout.add_widget(self.input2)
        self.input2_layout.opacity = 0  # Hidden initially
        self.input_layout.add_widget(self.input2_layout)

        self.add_widget(self.input_layout)

        # Convert button
        self.convert_btn = Button(
            text='Convert',
            size_hint=(1, None),
            height=50,
            background_color=(0.2, 0.8, 0.4, 1)
        )
        self.convert_btn.bind(on_press=self.convert)
        self.add_widget(self.convert_btn)

        # Result display
        self.result_label = Label(
            text='Result will appear here',
            font_size='18sp',
            size_hint_y=None,
            height=60,
            color=(0.1, 0.1, 0.1, 1),
            text_size=(Window.width - 40, None)
        )
        self.add_widget(self.result_label)

        # Clear button
        clear_btn = Button(
            text='Clear All',
            size_hint=(1, None),
            height=40,
            background_color=(0.8, 0.2, 0.2, 1)
        )
        clear_btn.bind(on_press=self.clear_fields)
        self.add_widget(clear_btn)

        # Set initial operation
        self.current_operation = None

    def on_operation_change(self, spinner, text):
        """Handle operation selection change"""
        self.current_operation = text

        # Show/hide second input based on operation
        if text == 'Binary Addition':
            self.input2_layout.opacity = 1
            self.input2_layout.disabled = False
            self.input1.hint_text = 'Enter first binary number'
            self.input2.hint_text = 'Enter second binary number'
        else:
            self.input2_layout.opacity = 0
            self.input2_layout.disabled = True
            if text == 'Decimal to Binary':
                self.input1.hint_text = 'Enter decimal number'
            elif text == 'Binary to Octal':
                self.input1.hint_text = 'Enter binary number'
            elif text == 'Binary to Hexadecimal':
                self.input1.hint_text = 'Enter binary number'

    def convert(self, instance):
        """Perform the selected conversion"""
        if not self.current_operation or self.current_operation == 'Select Operation':
            self.show_popup('Error', 'Please select an operation first!')
            return

        converter = BinaryConverter()
        result = ""

        try:
            if self.current_operation == 'Decimal to Binary':
                if not self.input1.text:
                    self.show_popup('Error', 'Please enter a decimal number!')
                    return
                result = converter.decimal_to_binary(int(self.input1.text))

            elif self.current_operation == 'Binary Addition':
                if not self.input1.text or not self.input2.text:
                    self.show_popup('Error', 'Please enter both binary numbers!')
                    return
                result = converter.binary_addition(self.input1.text, self.input2.text)

            elif self.current_operation == 'Binary to Octal':
                if not self.input1.text:
                    self.show_popup('Error', 'Please enter a binary number!')
                    return
                result = converter.binary_to_octal(self.input1.text)

            elif self.current_operation == 'Binary to Hexadecimal':
                if not self.input1.text:
                    self.show_popup('Error', 'Please enter a binary number!')
                    return
                result = converter.binary_to_hex(self.input1.text)

            self.result_label.text = f'Result: {result}'

        except Exception as e:
            self.show_popup('Error', f'An error occurred: {str(e)}')

    def clear_fields(self, instance):
        """Clear all input fields and results"""
        self.input1.text = ''
        self.input2.text = ''
        self.result_label.text = 'Result will appear here'
        self.operation_spinner.text = 'Select Operation'
        self.input2_layout.opacity = 0
        self.input2_layout.disabled = True

    def show_popup(self, title, message):
        """Show a popup message"""
        popup_layout = BoxLayout(orientation='vertical', padding=10)
        popup_layout.add_widget(Label(text=message))

        close_btn = Button(text='OK', size_hint=(1, None), height=40)
        popup = Popup(title=title, content=popup_layout, size_hint=(0.8, 0.4))

        close_btn.bind(on_press=popup.dismiss)
        popup_layout.add_widget(close_btn)

        popup.open()

class BinaryConverterApp(App):
    """Main application class"""

    def build(self):
        self.title = 'Binary Converter Pro'
        return BinaryConverterLayout()

if __name__ == '__main__':
    # Set window size for desktop testing
    Window.size = (400, 600)
    BinaryConverterApp().run()
```
