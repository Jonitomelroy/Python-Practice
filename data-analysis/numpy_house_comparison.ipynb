```python
import numpy as np

house_A = np.array([1500, 3, 10])   # size(sqft), bedrooms, age(years)
house_B = np.array([1600, 3, 8])
house_C = np.array([800, 1, 40])

# Task: compute the dot product of A with B, and A with C
# Which pair is "more similar" based on the dot product?
sum_a_b,sum_a_c=0,0
for i in range(len(house_A)):
   sum_a_b=int(house_A[i])*int(house_B[i])+sum_a_b
   sum_a_c=int(house_A[i])*int(house_C[i])+sum_a_c
if sum_a_b <sum_a_c:
  print("house_A and house_B are more similar")
else:
  print("house_A and house_C are more similar")
```
