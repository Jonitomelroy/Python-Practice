```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=sns.load_dataset('titanic')
plt.figure(figsize=(10,5))
sns.histplot(data=df , x='class')
plt.xlabel('Class')
plt.ylabel('Frequency')
plt.title('Distribution of Class')
plt.grid(True)
plt.show()
```

```python
df=sns.load_dataset('titanic')
plt.figure(figsize=(10,5))
sns.histplot(data=df , x='fare' , bins=100 , color="blue")
plt.xlabel('fare')
plt.ylabel('Frequency')
plt.title('Distribution of fare')
plt.grid(True)
plt.show()
```

```python
df=sns.load_dataset('titanic')
plt.figure(figsize=(10,5))
sns.histplot(data=df , x='fare' , bins=70 , color="blue" ,kde=True ,alpha=0.1)
plt.xlabel('fare')
plt.ylabel('Frequency')
plt.title('Distribution of fare')
plt.grid(True)
plt.show()
```

```python
df=sns.load_dataset('titanic')
plt.figure(figsize=(10,5))
sns.histplot(data=df , x='class' , bins=70 , color="blue",hue="adult_male" ,kde=True ,alpha=0.5)
plt.xlabel('class')
plt.ylabel('Frequency')
plt.title('Distribution of class based on gender')
plt.grid(True)
plt.show()
```
