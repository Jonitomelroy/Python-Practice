```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Create a figure and axes for the subplots
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Plotting Quantity vs Product
sns.barplot(x='Product', y='Quantity', data=df, ax=axes[0])
axes[0].set_title('Quantity by Product')

# Plotting Unit Price vs Product
sns.barplot(x='Product', y='Unit Price', data=df, ax=axes[1])
axes[1].set_title('Unit Price by Product')

# Plotting Total vs Product
sns.barplot(x='Product', y='Total', data=df, ax=axes[2])
axes[2].set_title('Total by Product')

plt.tight_layout()
plt.show()
```
